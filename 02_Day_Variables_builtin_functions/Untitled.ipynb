## Variables in Python 

- Here I revisited the idea of creating variables in python. 
- Remember a variable is a labeled container that stores a value we'd like to reuse in the future. 

```python
# Declaring the variables in Python 

first_name = "Janine" 
last_name = "Makorre"
country = "Tanzania"
city = "Mwanza"
age = 30 
is_married = False

skills = ["Python","R", "HTML", "C#", "C++", "Java"] # variable is a list 
personal_info = {# Variable is a dictionary
    "first name":"Janine",
    "last name":"Makorre",
    "country" : "Tanzania",
    "city" : "Mwanza"} 

# printing the variables 
print(f'Name: {first_name} {last_name}')
print(f'Country and city: {country} - {city}')
print (f"Age: {age}")
print(f"Marital Status: {is_married}")
```

## Checking for data types 

In python, you check for data types using an inbuilt function type(). 

```python
# checking the data types 

print(type(first_name))

print(type(age))

print(type(is_married))

print(type(skills))

print(type(personal_info))
```

```python
# counting the number of characters in a string 

print('First name length:', len(first_name))

# counting the number of items in a list 
print(f"Number of Items in list: {len(skills)}")
```

```python
# comparing the length of my first name and last name 

print(f"length of first name: {len(first_name)} vs length of last name: {len(last_name)}")
```

```python
print(max(len(first_name), len(last_name)))
```

```python
# Exercise 
num_one = 5 
num_two = 4 

# Computations 
total = num_one + num_two
diff = num_one - num_two
prod = num_one * num_two
division = num_one / num_two
remainder = num_two % num_one
exponential = num_one**num_two
floor_division = num_one // num_two

# printing them 
print(f"total: {total}")
print(f"diff: {diff}")
print(f"prod: {prod}")
print(f"division: {division}")
print(f"remainder: {remainder}")
print(f"exponential: {exponential}")
print(f"floor_division: {floor_division}")
```

**Circle calculations**

The radius of a circle is 30 meters.

i) Calculate the area of a circle and assign the value to a variable name of area_of_circle

ii) Calculate the circumference of a circle and assign the value to a variable name of circum_of_circle

iii) Take radius as user input and calculate the area.

```python
r = 30
pi = 3.14
area_of_circle = pi*r**2
print(f"Area of circle: {area_of_circle} m\u00b2")
```

```python
import math 
r = 30 
area_of_circle = math.pi * r**2
area_of_circle
```
